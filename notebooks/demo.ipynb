# OBELiX External Datasets Demo

This notebook demonstrates how to load, explore, and combine the external solid-state electrolyte (SSE) datasets available in the OBELiX package:

| Class | Source | Description |
|-------|--------|-------------|
| `LiIon` | [Morgan et al.](https://pcwww.liv.ac.uk/~msd30/lmds/) | Liverpool Materials Discovery Server lithium-ion conductivity database |
| `Laskowski` | [Laskowski et al.](https://doi.org/10.1038/s41560-023-01401-0) | Room-temperature ionic conductivities with ICSD references |
| `ShonAndMin` | [Shon & Min (2023)](https://doi.org/10.1021/acsomega.3c01424) | ACS Omega supplementary ionic conductivity data |
| `McHaffie` | [McHaffie et al. (2025)](https://doi.org/10.22002/a3fhb-wyb45) | Caltech Data repository: 571 Li-ion conductivity entries with ICSD codes |

For the full OBELiX dataset and paper, see: https://arxiv.org/html/2502.14234v2

## Setup

```python
import obelix
from obelix import LiIon, Laskowski, ShonAndMin, McHaffie, Dataset

print(f"OBELiX version: {obelix.__version__}")
```

## LiIon Dataset

The LiIon dataset comes from the Liverpool Materials Discovery Server. It contains lithium-ion conductivity measurements across a wide range of compositions and crystal families.

By default, only room-temperature measurements (25 &plusmn; 7 &deg;C) are kept.

```python
liion = LiIon()
print(f"LiIon: {len(liion)} entries")
print(f"Columns: {liion.dataframe.columns.tolist()}")
```

```python
liion.dataframe.head(10)
```

```python
liion.dataframe['Ionic conductivity (S cm-1)'].describe()
```

## Laskowski Dataset

The Laskowski dataset contains room-temperature ionic conductivity values with corresponding ICSD entries and DOIs. The data has been curated with manually verified DOIs.

```python
laskowski = Laskowski(fast=True)
print(f"Laskowski: {len(laskowski)} entries")
print(f"Columns: {laskowski.dataframe.columns.tolist()}")
```

```python
laskowski.dataframe.head(10)
```

```python
laskowski.dataframe['Ionic conductivity (S cm-1)'].describe()
```

## ShonAndMin Dataset

The Shon & Min dataset is extracted from the supplementary information of their ACS Omega (2023) paper. The raw data undergoes several cleaning steps:

1. **Ionic conductivity parsing** &mdash; human-written values (ranges, varied units) are normalized to S/cm
2. **Formula validation** &mdash; compositions are validated with pymatgen
3. **Minimum conductivity** &mdash; for each (composition, DOI) pair, only the minimum conductivity is kept
4. **Room temperature filter** &mdash; only 25 &plusmn; 7 &deg;C measurements are retained

```python
shonandmin = ShonAndMin()
print(f"ShonAndMin: {len(shonandmin)} entries")
print(f"Columns: {shonandmin.dataframe.columns.tolist()}")
```

```python
shonandmin.dataframe.head(10)
```

```python
shonandmin.dataframe['Ionic conductivity (S cm-1)'].describe()
```

## McHaffie Dataset

The McHaffie dataset is sourced from the Caltech Data repository (McHaffie et al., 2025). It contains 571 lithium-ion conductivity entries, each with an ICSD collection code and DOI. Because ICSD structural data requires a licence, the `get_space_groups()`, `get_lattice_parameters()`, and `get_cifs()` methods raise `NotImplementedError` as stubs for future integration.

```python
mchaffie = McHaffie()
print(f"McHaffie: {len(mchaffie)} entries")
print(f"Columns: {mchaffie.dataframe.columns.tolist()}")
```

```python
mchaffie.dataframe.head(10)
```

```python
mchaffie.dataframe['Ionic conductivity (S cm-1)'].describe()
```

## Combining Datasets with `+`

The `+` operator concatenates two `Dataset` objects, keeping only columns that are common to both. No deduplication is performed &mdash; all rows from both datasets are included.

```python
combined = liion + laskowski

print(f"LiIon:      {len(liion)} entries")
print(f"Laskowski:  {len(laskowski)} entries")
print(f"Combined:   {len(combined)} entries  (sum = {len(liion) + len(laskowski)})")
print(f"\nCommon columns: {combined.dataframe.columns.tolist()}")
```

```python
combined.dataframe.head()
```

You can chain additions to combine all four datasets at once:

```python
all_external = liion + laskowski + shonandmin + mchaffie

print(f"All external datasets: {len(all_external)} entries")
print(f"Columns: {all_external.dataframe.columns.tolist()}")
```

## Deduplication with `union`

The `union` method works like `+` but removes rows whose reduced composition (via pymatgen's canonical formula) duplicates an earlier row. This is useful when the same material appears in multiple datasets.

```python
unioned = liion.union(laskowski)

naive_total = len(liion) + len(laskowski)
print(f"LiIon + Laskowski (with duplicates): {naive_total} entries")
print(f"LiIon.union(Laskowski):              {len(unioned)} entries")
print(f"Duplicates removed:                  {naive_total - len(unioned)}")
```

## Merging with `merge_datasets`

`Dataset.merge_datasets()` is a static method that merges any number of datasets. It restricts the output to a curated set of columns (`Reduced Composition`, `Ionic conductivity (S cm-1)`, `Space group #`, `DOI`) &mdash; only those present in **all** input datasets are kept.

By default, duplicates are removed by canonical reduced composition.

```python
merged = Dataset.merge_datasets(liion, laskowski, shonandmin, mchaffie)
merged_with_dups = Dataset.merge_datasets(liion, laskowski, shonandmin, mchaffie, remove_duplicates=False)

print(f"Merged (deduplicated):    {len(merged)} entries")
print(f"Merged (with duplicates): {len(merged_with_dups)} entries")
print(f"Duplicates removed:       {len(merged_with_dups) - len(merged)}")
print(f"\nColumns: {merged.dataframe.columns.tolist()}")
```

```python
merged.dataframe.head(10)
```

## Cross-Dataset Deduplication with `remove_matching_entries`

When combining data from different sources, entries often overlap. The `remove_matching_entries` method identifies and removes duplicates based on:

1. **Composition match** &mdash; canonical reduced formula equivalence (e.g., `Li2O` and `Li4O2` are treated as the same)
2. **DOI match** &mdash; same DOI confirms the measurement comes from the same paper
3. **Missing DOI** &mdash; if the entry has no DOI but the composition exists in the other dataset with a known DOI, it is also removed

Entries with the same composition but *different* DOIs are kept &mdash; they represent distinct measurements from different papers.

```python
liion_cleaned = liion.remove_matching_entries(laskowski)

print(f"LiIon before:           {len(liion)} entries")
print(f"LiIon after dedup:      {len(liion_cleaned)} entries")
print(f"Entries removed:        {len(liion) - len(liion_cleaned)}")
```

```python
shon_cleaned = shonandmin.remove_matching_entries(laskowski)

print(f"ShonAndMin before:      {len(shonandmin)} entries")
print(f"ShonAndMin after dedup: {len(shon_cleaned)} entries")
print(f"Entries removed:        {len(shonandmin) - len(shon_cleaned)}")
```

Note that the original datasets are never modified &mdash; `remove_matching_entries` always returns a new `Dataset`:

```python
print(f"Original LiIon still has {len(liion)} entries (unchanged)")
```

|-------|---------|-------------|--------|
| `LiIon` | See above | Reduced Composition, Ionic conductivity, DOI, Family | Liverpool Materials Discovery Server |
| `Laskowski` | See above | Reduced Composition, Ionic conductivity, DOI, Space group, ICSD | Laskowski et al. (2023) |

| `McHaffie` | See above | Reduced Composition, Ionic conductivity, DOI, ICSD Collection Code | McHaffie et al. (2025) |

**Key operations:**

| Operation | Method | Deduplicates? | Notes |
|-----------|--------|---------------|-------|
| Concatenate | `ds1 + ds2` | No | Keeps common columns, all rows |
| Union | `ds1.union(ds2)` | By composition | Removes formula-equivalent duplicates |
| Merge | `Dataset.merge_datasets(ds1, ds2, ...)` | By composition (default) | Restricts to curated column set |
| Remove overlaps | `ds1.remove_matching_entries(ds2)` | By composition + DOI | Keeps entries with different DOIs |
